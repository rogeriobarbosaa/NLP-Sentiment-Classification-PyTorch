```python
from src.LSTM import LSTM
```

### Treinamento

Pipeline de Treino:
1. Modelo
2. Loss e Otimizador
3. Loop de Treino:
    - Forward Pass
    - Calcula Loss
    - Backward Pass
    - Atualiza Pesos

### 0. Preparação

```python
X_train = torch.load("data/X_train.pt")
y_train = torch.load("data/y_train.pt")
```

```python
dataset = TensorDataset(X_train, y_train)
dataset
```

```python
train_loader = DataLoader(dataset=dataset, batch_size=30, shuffle=True)
train_loader
```

### 1. Modelo

```python
input_size = 100
hidden_size = 5
num_layers = 2
num_classes = 3
```

```python
model = LSTM(input_size, hidden_size, num_layers, num_classes)
```

### 2. Loss e Otimizador

```python
learning_rate = 0.002
```

```python
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(params=model.parameters(), lr=learning_rate)
```

### 3. Treinamento

```python
num_epochs = 50
```

```python
for epoch in range(num_epochs):
    epoch_loss = 0

    for i, (batch_feat, batch_label) in enumerate(train_loader):  # itera a cada batch
        # Forward pass
        y_pred = model(batch_feat)  # previsões do batch atual

        # Calcula loss
        loss = criterion(y_pred, batch_label)  # loss do batch atual
        epoch_loss += loss

        # Backward pass
        loss.backward()

        # Updates
        optimizer.step()  # atualiza o otimizador
        optimizer.zero_grad()

    print(f"Epoch: {epoch+1} | Loss = {epoch_loss/(i+1)}")
```

### Salvando o modelo

```python
torch.save(model, 'data/model.pth')
```
