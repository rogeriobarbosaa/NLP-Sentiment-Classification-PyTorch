```python
import os
import torch
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
```

```python
os.chdir("../")
```

### Avaliação

```python
X_test = torch.load("data/X_test.pt")
y_test = torch.load("data/y_test.pt")
```

```python
model = torch.load("data/model.pth", weights_only=False)
```

```python
with torch.no_grad():
    y_pred = model(X_test)
    y_pred = torch.argmax(y_pred, dim=1)  # pega a saída de maior probabilidade, dentre as saídas da rede (como o problema é ternário, temos 3 saídas na rede)

    acuracia = accuracy_score(y_test, y_pred)
    precisao = precision_score(y_test, y_pred, average='weighted')
    recall = recall_score(y_test, y_pred, average='weighted')
    f1 = f1_score(y_test, y_pred, average='weighted')

    print(f'acurácia = {acuracia:.4f}\nprecisão = {precisao:.4f}\nrecall = {recall:.4f}\nf1 score = {f1:.4f}')
```
